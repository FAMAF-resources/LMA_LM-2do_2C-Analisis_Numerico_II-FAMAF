```python
import numpy as np
import time

def det_lu(A):
    n = A.shape[0]
    U = A.copy()
    P = np.eye(n)
    detP = 1
    for k in range(n):
        l = k + np.argmax(np.abs(U[k:, k]))
        if l != k:
            detP = detP * (-1)
            U[[k, l], :] = U[[l, k], :]

            P[[k, l], :] = P[[l, k], :]

        U[k+1:, k] = U[k+1:, k] / U[k, k]

        U[k+1:, k+1:] = U[k+1:, k+1:] - np.outer(U[k+1:, k], U[k, k+1:])

    L = np.tril(U, -1) + np.eye(n)
    U = np.triu(U)

    diag_U = np.diag(U)
    det_U = 1
    det_L = 1
    det_P = detP
    for i in range (n):
        det_U = det_U * diag_U[i]

    det_A = det_P * det_U * det_L

    return det_A

A = np.array([[2., 10, 8, 8, 6], [1, 4, -2, 4, -1], [0, 2, 3, 2, 1], [3, 8, 3, 10, 9], [1, 4, 1, 2, 1]])

inicio = time.time()
det_A = det_lu(A)
fin = time.time()

inicio_1 = time.time()
det = np.linalg.det(A)
fin_1 = time.time()

print(f'El determinande calculado con det_lu es {det_A}')
print(f'su tiempo de ejecucion es {fin-inicio}')

print(f'El determinande calculado con np.linalg.det es {det}')
print(f'su tiempo de ejecucion es {fin_1-inicio_1}')
```
