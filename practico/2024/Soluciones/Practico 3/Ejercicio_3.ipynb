```python
# Importamos las librerias que vamos a necesitar
import numpy as np
import matplotlib.pyplot as plt
```

```python
n_puntos = 1000

# Generamos 4 array de ceros para guardar nuestros pares (x,y) tales que ||(x,y)||_p<=1 (bola unidad)
vectores_1 = np.zeros((n_puntos,2))
vectores_2 = np.zeros((n_puntos,2))
vectores_3 = np.zeros((n_puntos,2))
vectores_4 = np.zeros((n_puntos,2))
```

```python
# Norma 1
k = 0
while k < n_puntos: # Usamos while hasta completar todos los vectores
    xy = np.random.random(2) # Generamos un vector random (x, y)
    norm_xy = np.linalg.norm(xy, 1)
    if norm_xy <= 1: # vemos si (x,y) se encuentra en la bola unidad
        vectores_1[k,:] = np.array([xy[0], xy[1]]) # (x, y)
        vectores_2[k,:] = np.array([-xy[0], -xy[1]]) # (-x, -y)
        vectores_3[k,:] = np.array([-xy[0], xy[1]]) # (-x, y)
        vectores_4[k,:] = np.array([xy[0], -xy[1]]) # (x, -y)
        k = k+1

block_vector = np.block([[vectores_1], [vectores_2], [vectores_3], [vectores_4]]) # Juntamos todo en un bloque

# Graficamos
plt.plot(block_vector[:, 0], block_vector[:, 1], '*', 'b')
plt.axis('equal')
plt.show()
```

```python
# Norma 2
k = 0
while k < n_puntos:
    xy = np.random.random(2)
    norm_xy = np.linalg.norm(xy, 2)
    if norm_xy <= 1:
        vectores_1[k,:] = np.array([xy[0], xy[1]])
        vectores_2[k,:] = np.array([-xy[0], -xy[1]])
        vectores_3[k,:] = np.array([-xy[0], xy[1]])
        vectores_4[k,:] = np.array([xy[0], -xy[1]])
        k = k+1

block_vector = np.block([[vectores_1], [vectores_2], [vectores_3], [vectores_4]])

plt.plot(block_vector[:, 0], block_vector[:, 1], '*', 'b')
plt.axis('equal')
plt.show()
```

```python
# Norma infinito
k = 0
while k < n_puntos:
    xy = np.random.random(2)
    norm_xy = np.linalg.norm(xy, np.inf)
    if norm_xy <= 1:
        vectores_1[k,:] = np.array([xy[0], xy[1]])
        vectores_2[k,:] = np.array([-xy[0], -xy[1]])
        vectores_3[k,:] = np.array([-xy[0], xy[1]])
        vectores_4[k,:] = np.array([xy[0], -xy[1]])
        k = k+1

block_vector = np.block([[vectores_1], [vectores_2], [vectores_3], [vectores_4]])

plt.plot(block_vector[:, 0], block_vector[:, 1], '*', 'b')
plt.axis('equal')
plt.show()
```
