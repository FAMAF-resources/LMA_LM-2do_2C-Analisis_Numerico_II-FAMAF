```python
# Una forma

import numpy as np

def cuad_min_svd(A, b):

    m, n = A.shape

    U, S, Vt = np.linalg.svd(A)

    V = Vt.T

    r = min(m,n)

    rango = np.where(np.abs(S) < 1e-10)[0] #Es mejor agregar valor absoluto
    if len(rango) > 0:
        r = rango[0]

    x_sol = np.zeros((n, m))

    for i in range(r):
       x_sol = x_sol + np.outer(V[:, i],U[:, i])/S[i]

    x_sol = x_sol@b

    min_svd = 0
    for i in range(r+1, m):
        min_svd = min_svd +(U[:, i].T@b)**2

    return x_sol, np.sqrt(min_svd)
```

```python
# Otra forma

import numpy as np

def cuad_min_svd(A, b):
    m, n = A.shape
    U, S, Vt = np.linalg.svd(A)
    b_tilde = U.T @ b
    r = min(m,n)
    rango = np.where(np.abs(S)< 1e-15)[0]
    if len(rango) > 0:
        r = rango[0]
    y = b_tilde[:r] / S[:r]
    x_tilde = np.zeros(n)
    x_tilde[:r] = y

    x = Vt.T @ x_tilde
    return x, np.linalg.norm(b - A @ x, 2)
```

```python
x_sol, min_svd = cuad_min_svd(A_datos, b_datos)
print(x_sol, min_svd)
```
