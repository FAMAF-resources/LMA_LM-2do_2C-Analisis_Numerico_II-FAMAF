```python
from matplotlib import pyplot as plt
import numpy as np
import sys


A_datos = np.loadtxt("/content/drive/MyDrive/Ejercicios resueltos P5/p5e5.txt")

#Aproximar imagenes mediante datos
# usa el comando imshow para mostrar las imagenes que forman las matrices

def im_aprox_svd(A, tol):
    m, n = A.shape
    u, s, vt = np.linalg.svd(A)

    for k in range(min(m,n)):
        u_k = u[:, :k]
        s_k = np.diag(s)[:k, :k]
        vt_k = vt[:k, :]
        A_k = u_k@s_k@vt_k
        norm = np.linalg.norm(A-A_k, 2)
        if norm < tol:
            break

    fig, ((ax1, ax2)) = plt.subplots(1,2)

    ax1.imshow(A_k)
    ax1.title.set_text('A_k')
    ax1.axis('off')
    ax2.imshow(A)
    ax2.title.set_text('A')
    ax2.axis('off')

    plt.show()

sol = im_aprox_svd(A_datos, 2000)
```
