```python
import numpy as np

import sys
sys.path.append('Practico1')
sys.path.append('Practico2')

#Funciones que van a necesitar
from Ejercicio_6_dlu import dlu
from Ejercicio_3__sol_trisup import sol_trsupfil
from Ejercicio_3__sol_trinf import sol_trinffil
from sol_cuadim import sol_cuadmin

n = 100

def sol_ec_normal(n):

    b = np.array([1.,1,1])
    epsilon = np.zeros(n)
    sol = np.zeros((2, n))
    cond_normal = np.zeros(n)
    for i in range(n):
        eps = 2**(-(i+1))
        A = np.array([[1., 1],[eps, 0], [0, eps]])
        A_normal = A.T@A
        kappa = np.linalg.cond(A_normal)
        b_normal = A.T@b
        L, U  = dlu(A_normal)
        y = sol_trinffil(L, b_normal)
        x = sol_trsupfil(U, y)

        #Guardamos todo
        cond_normal[i] = kappa
        epsilon[i] = eps
        sol[:, i] = x

    return sol, epsilon, cond_normal

sol_nor, epsilon_nor, cond = sol_ec_normal(n)
print(f'sol_nor = {sol_nor[:, 0]} y num_cond = {cond[0]}')

def sol_qr_cuadmin(n):
    b = np.array([1.,1,1])
    epsilon = np.zeros(n)
    sol = np.zeros((2, n))
    cond_A = np.zeros(n)
    for i in range(n):
        eps = 2**(-(i+1))
        A = np.array([[1., 1],[eps, 0], [0, eps]])
        kappa = np.linalg.cond(A)
        x, res = sol_cuadmin(A, b)
        epsilon[i] = eps
        sol[:, i] = x
        cond_A[i] = kappa

    return sol, epsilon, cond_A

n = 100
sol_qr, epsilon_qr, cond_qr = sol_qr_cuadmin(n)
print(f'sol_qr = {sol_qr[:, 0]}, cond = {cond_qr[0]}')
```
