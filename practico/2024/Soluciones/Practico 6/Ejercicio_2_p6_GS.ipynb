```python
import numpy as np

# Metodo Gauss-Seidel

def sol_gseidel(A, b, x_0, err, M):
    n = A.shape[0]
    for l in range(n):
        if A[l, l] == 0:
            print("hay al menos un cero en la diagonal")
            return None
    D = np.diag(np.diag(A))
    U = np.triu(A,1)
    L = np.tril(A, -1)
    D_inv = np.zeros((n,n))
    for j in range(n):
        D_inv[j,j] = 1/D[j,j]

    b = D_inv @ b
    A = D_inv @ (L + U)

    x_mas = x_0

    for k in range(M):
        x_tilde = x_mas.copy()
        for i in range(n):
            x_mas[i] = b[i] - A[i, :] @ x_tilde

        norm = np.linalg.norm(x_mas - x_tilde, np.inf)
        if norm <= err:
            print("llegamos a la solucion con", k)
            break

    return x_mas

A = np.array([[4., 1],
              [2, 3]])
b = np.array([7., 5])

x0 = np.array([0., 0])

x_sol = sol_gseidel(A, b, x0, 1e-10, 100)
print(b- A@x_sol)
```
