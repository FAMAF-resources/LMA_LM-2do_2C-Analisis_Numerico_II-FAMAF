```python
import numpy as np

# Método de Gastinel

def sol_gastinel(A, b, x_0, err, M):
    n = A.shape[0]
    r = b - A @ x_0
    sigma = np.linalg.norm(r, np.inf)
    d = np.sign(r)
    x_mas = x_0.copy()
    for k in range(M):
        if sigma <= err:
            print("llegamos a la aproximacion", k)
            break
        else:
            alpha = np.dot(r, d) / np.dot(d, A @ d)
            x_mas = x_mas + alpha * d
            r = r - alpha * (A@d)
            d = np.sign(r)
            sigma = np.linalg.norm(r, np.inf)

    return x_mas, sigma

A = np.random.random((3,3))
A = 0.5*(A+A.T)
A = np.eye(3)+A
b = np.random.random(3)
x_0 = np.zeros(3)
x_sol, sigma = sol_gastinel(A, b, x_0, 1e-10, 500)
print(A@x_sol - b)
```
